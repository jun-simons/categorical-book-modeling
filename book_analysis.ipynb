```python
import seaborn as sns
from sklearn.feature_extraction.text import CountVectorizer

sns.set(style="whitegrid")
plt.rcParams["figure.figsize"] = (8, 5)
text_df = books_df.dropna(subset=["Description", "Rating"]).copy()
text_df = text_df[(text_df["Rating"] >= 1.0) & (text_df["Rating"] <= 5.0)]

print("Usable rows with description & rating:", text_df.shape)
max_rows = 1_000_000
if len(text_df) > max_rows:
    text_df = text_df.sample(max_rows, random_state=42)
    print("Sampled down to", len(text_df), "rows for text EDA.")

# Binary label: high vs low rating
rating_threshold = 4.0
text_df["high_rating"] = (text_df["Rating"] >= rating_threshold).astype(int)

# Rating bins for nicer plots
text_df["rating_bin"] = pd.cut(
    text_df["Rating"],
    bins=[0, 3.5, 4.0, 4.5, 5.1],
    labels=["≤3.5", "3.5–4.0", "4.0–4.5", ">4.5"]
)

# 2. Length features: characters and words
text_df["desc_len_chars"] = text_df["Description"].str.len()
text_df["desc_len_words"] = text_df["Description"].str.split().apply(len)

print("\nLength summary (chars):")
print(text_df["desc_len_chars"].describe())
print("\nLength summary (words):")
print(text_df["desc_len_words"].describe())

# Clip lengths for nicer plots
text_df["desc_len_words_clipped"] = text_df["desc_len_words"].clip(upper=500)
# 3. Plots: length distributions overall and by rating
# (a) Overall distribution of description length (words)
plt.figure()
sns.histplot(text_df["desc_len_words_clipped"], bins=40)
plt.title("Distribution of Description Length (words, clipped at 500)")
plt.xlabel("Description length (words)")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
# (b) Length distribution by high vs low ratinfg
plt.figure()
sns.kdeplot(
    data=text_df,
    x="desc_len_words_clipped",
    hue="high_rating",
    common_norm=False,
    fill=True,
    alpha=0.4
)
plt.title("Description Length by Rating Group")
plt.xlabel("Description length (words, clipped at 500)")
plt.legend(title="High rating (>= 4.0)")
plt.tight_layout()
plt.show()
# (c) Boxplot: length vs rating bin
plt.figure()
sns.boxplot(
    data=text_df,
    x="rating_bin",
    y="desc_len_words",
    showfliers=False
)
plt.title("Description Length vs Rating Bin")
plt.xlabel("Rating bin")
plt.ylabel("Description length (words)")
plt.tight_layout()
plt.show()
# (d) Scatter: rating vs length  of a subsample for the scatter plot
plt.figure()
scatter_sample = text_df.sample(min(10_000, len(text_df)), random_state=42)
sns.scatterplot(
    data=scatter_sample,
    x="desc_len_words",
    y="Rating",
    alpha=0.3
)
plt.title("Rating vs Description Length (sample)")
plt.xlabel("Description length (words)")
plt.ylabel("Average rating")
plt.tight_layout()
plt.show()

# 4. Content analysis: common words in high vs low rated bfooks
content_sample = text_df.sample(min(200_000, len(text_df)), random_state=123)

vectorizer = CountVectorizer(
    max_features=5000,
    min_df=50,
    stop_words="english"
)
X = vectorizer.fit_transform(content_sample["Description"].astype(str))
vocab = np.array(vectorizer.get_feature_names_out())

mask_high = content_sample["high_rating"].values == 1
mask_low = ~mask_high

# Sum counts per token in each group
counts_high = np.asarray(X[mask_high].sum(axis=0)).ravel()
counts_low = np.asarray(X[mask_low].sum(axis=0)).ravel()

# Normalize to get relative frequencies
freq_high = counts_high / counts_high.sum()
freq_low = counts_low / counts_low.sum()

# Compute a simple "lift"
eps = 1e-9
lift = (freq_high + eps) / (freq_low + eps)

# Top words for high-rated books
min_freq = 1e-4  # ignore super-rare words
valid = freq_high > min_freq

top_high_idx = np.argsort(-lift * valid)[:20]
top_high_words = pd.DataFrame({
    "token": vocab[top_high_idx],
    "freq_high": freq_high[top_high_idx],
    "freq_low": freq_low[top_high_idx],
    "lift_high_over_low": lift[top_high_idx]
})

print("\nTop tokens associated with HIGH-rated books (by lift):")
print(top_high_words.to_string(index=False))

# Top words for low-rated books (inverse lift)
lift_low_over_high = (freq_low + eps) / (freq_high + eps)
valid_low = freq_low > min_freq
top_low_idx = np.argsort(-lift_low_over_high * valid_low)[:20]
top_low_words = pd.DataFrame({
    "token": vocab[top_low_idx],
    "freq_high": freq_high[top_low_idx],
    "freq_low": freq_low[top_low_idx],
    "lift_low_over_high": lift_low_over_high[top_low_idx]
})

print("\nTop tokens associated with LOWER-rated books (by lift):")
print(top_low_words.to_string(index=False))
```

## Planning: basic recommender system

Problem formulation: we have a user-book matrix with an entry of 1 if the user marked that book as read/rated and 0 otherwise.

On the simplest level, this can be treated as a clustering problem:
We have a very high dimensional vector space (corresponding with the number of books)
Each user is a vector in R^M where M is the number of books
Run an algorithm such as K-means or K-medoids to group users whose vectors are close

Similarity metric:
Euclidean distance, or other standard distances are poor choices for these high dimensional vector spaces.  Jaccard similarity, defined as IOU over book iinteractions, is a standard choice.  We could also take the cosine distance between vectors in this space, which is inherently normalized by the users reading history.

Practical notes:
we should probably use scipy.sparse or a similar more efficient library (can allow for more efficient algorithms for cosine similarity, distance, SVD, etc)

```python
import numpy as np
import pandas as pd
from scipy.sparse import csr_matrix

interactions = merged.drop_duplicates(subset=["ID", "Id"])
unique_users = interactions["ID"].unique()
unique_books = interactions["Id"].unique()

user_id_to_idx = {uid: i for i, uid in enumerate(unique_users)}
book_id_to_idx = {bid: j for j, bid in enumerate(unique_books)}

interactions["user_idx"] = interactions["ID"].map(user_id_to_idx)
interactions["book_idx"] = interactions["Id"].map(book_id_to_idx)

n_users = len(unique_users)
n_books = len(unique_books)

# Build full sparse matrix
row = interactions["user_idx"].to_numpy()
col = interactions["book_idx"].to_numpy()
data = np.ones(len(interactions), dtype=np.uint8)

X = csr_matrix((data, (row, col)), shape=(n_users, n_books))

# Filter active users/books
min_ratings_per_user = 20
min_users_per_book = 20

user_counts = np.asarray(X.sum(axis=1)).ravel()
book_counts = np.asarray(X.sum(axis=0)).ravel()

user_mask = user_counts >= min_ratings_per_user
book_mask = book_counts >= min_users_per_book

X_filtered = X[user_mask][:, book_mask]

print("Filtered matrix shape:", X_filtered.shape)

# Keep mapping from filtered indices back to original ids
filtered_user_indices = np.where(user_mask)[0]
filtered_book_indices = np.where(book_mask)[0]

idx_to_user_id = {i_new: unique_users[i_old] for i_new, i_old in enumerate(filtered_user_indices)}
idx_to_book_id = {j_new: unique_books[j_old] for j_new, j_old in enumerate(filtered_book_indices)}
```

```python
from collections import defaultdict
rng = np.random.default_rng(42)

# Build train/test interaction sets as dict
train_items = defaultdict(set)
test_items = defaultdict(set)

n_u, n_b = X_filtered.shape

# TRAIN TEST SPLITTING (try 80 20?)
test_frac = 0.2
min_test_items = 1

for u in range(n_u):
    # all indices of books the user interacted iwith
    start, end = X_filtered.indptr[u], X_filtered.indptr[u+1]
    user_books = X_filtered.indices[start:end]
    if len(user_books) == 0:
        continue

    # frac of test itesm
    n_test = max(min_test_items, int(len(user_books) * test_frac))
    n_test = min(n_test, len(user_books) - 1)  # keep at least 1 train if possible
    if n_test <= 0:
        train_items[u].update(user_books)
        continue

    test_idx = rng.choice(user_books, size=n_test, replace=False)
    test_set = set(test_idx)
    train_set = set(user_books) - test_set

    if len(train_set) == 0:
        # it all in test, not useful, so just put back in trrain
        b = test_set.pop()
        train_set.add(b)
        test_set.add(b)

    train_items[u].update(train_set)
    test_items[u].update(test_set)

# train only sparse matrix to use for SVD
train_row = []
train_col = []
for u, books in train_items.items():
    for b in books:
        train_row.append(u)
        train_col.append(b)

train_data = np.ones(len(train_row), dtype=np.uint8)
X_train = csr_matrix((train_data, (train_row, train_col)), shape=(n_u, n_b))
```

```python
from sklearn.decomposition import TruncatedSVD
from sklearn.preprocessing import normalize
from sklearn.cluster import KMeans

n_components = 50
svd = TruncatedSVD(n_components=n_components, random_state=42)
user_latent = svd.fit_transform(X_train)
user_latent = normalize(user_latent)

n_clusters = 5  
kmeans = KMeans(n_clusters=n_clusters, random_state=42, n_init="auto")
cluster_labels = kmeans.fit_predict(user_latent)
```

```python
# count how many users in that cluster have the book in TRAIN
cluster_book_counts = np.zeros((n_clusters, n_b), dtype=np.int32)

for u in range(n_u):
    c = cluster_labels[u]
    # books in train for this user
    books_u = train_items[u]
    for b in books_u:
        cluster_book_counts[c, b] += 1
```

```python
cluster_top_books = []
for c in range(n_clusters):
    counts = cluster_book_counts[c]
    # sort books by descending popularity
    order = np.argsort(-counts)
    cluster_top_books.append(order)
```

```python
def evaluate_cluster_recommender(K=10):
    hits = 0
    total_test_items = 0
    users_with_test = 0

    for u in range(n_u):
        test_set = test_items[u]
        if not test_set:
            continue
        users_with_test += 1
        total_test_items += len(test_set)

        c = cluster_labels[u]
        already_train = train_items[u]

        # recommend top-K books for this cluster that user hasn't seen in TRAIN
        recs = []
        for b in cluster_top_books[c]:
            if b in already_train:
                continue
            recs.append(b)
            if len(recs) >= K:
                break

        recs_set = set(recs)
        # count how many of the test items are in recs
        hits += len(test_set & recs_set)

    recall_at_k = hits / total_test_items if total_test_items > 0 else 0.0
    hit_rate = hits / users_with_test if users_with_test > 0 else 0.0

    return recall_at_k, hit_rate

for K in [5, 10, 20]:
    recall_k, hit_rate_k = evaluate_cluster_recommender(K=K)
    print(f"K={K:2d} | Recall@K={recall_k:.4f} | HitRate@K={hit_rate_k:.4f}")
```

```python
# Global popularity from TRAIN data
global_book_counts = np.zeros(n_b, dtype=np.int32)
for u in range(n_u):
    for b in train_items[u]:
        global_book_counts[b] += 1

global_top_books = np.argsort(-global_book_counts)

def evaluate_global_pop(K=10):
    hits = 0
    total_test_items = 0
    users_with_test = 0

    for u in range(n_u):
        test_set = test_items[u]
        if not test_set:
            continue
        users_with_test += 1
        total_test_items += len(test_set)

        already_train = train_items[u]
        recs = []
        for b in global_top_books:
            if b in already_train:
                continue
            recs.append(b)
            if len(recs) >= K:
                break

        recs_set = set(recs)
        hits += len(test_set & recs_set)

    recall_at_k = hits / total_test_items if total_test_items > 0 else 0.0
    hit_rate = hits / users_with_test if users_with_test > 0 else 0.0
    return recall_at_k, hit_rate

print("\nGlobal popularity baseline:")
for K in [5, 10, 20]:
    r, h = evaluate_global_pop(K)
    print(f"K={K:2d} | Recall@K={r:.4f} | HitRate@K={h:.4f}")
```

```python
import numpy as np
import pandas as pd

if "Description" not in books_df.columns:
    raise ValueError("books_df has no 'Description' column. Check the CSV or column name.")
# Keep only rows with non-null description and rating
text_df = books_df.dropna(subset=["Description", "Rating"]).copy()
text_df = text_df[(text_df["Rating"] >= 1.0) & (text_df["Rating"] <= 5.0)]

print("Usable rows with description & rating:", text_df.shape)
# in case dataset gets larger
max_rows = 1_000_000
if len(text_df) > max_rows:
    text_df = text_df.sample(max_rows, random_state=42)
    print("Sampled down to", len(text_df), "rows for modeling.")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline
from sklearn.metrics import classification_report, roc_auc_score

# Binary label, we use 1.0 if "highly rated", 0 else
rating_threshold = 4.0
y_class = (text_df["Rating"] >= rating_threshold).astype(int)
X_text = text_df["Description"].astype(str)

X_train, X_test, y_train, y_test = train_test_split(
    X_text, y_class,
    test_size=0.2,
    random_state=42,
    stratify=y_class
)

# Use TF–IDF for text, and run logistic regression
clf = make_pipeline(
    TfidfVectorizer(
        max_features=10_000,
        ngram_range=(1, 2),
        min_df=5,
        stop_words="english"
    ),
    LogisticRegression(
        max_iter=1000,
        n_jobs=-1,
        C = 0.5,
        solver="saga"
    )
)

print("Fitting classification model...")
clf.fit(X_train, y_train)

# Evaluate
y_proba = clf.predict_proba(X_test)[:, 1]
y_pred = (y_proba >= 0.5).astype(int)

print("\nClassification report (highly rated vs not):")
print(classification_report(y_test, y_pred, digits=3))

roc = roc_auc_score(y_test, y_proba)
print(f"ROC-AUC: {roc:.3f}")
```
